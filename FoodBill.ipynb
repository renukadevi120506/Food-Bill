```python
#Using Business Model with List ,Tuple,Dictionary
# Food Order

Food_item = ["Burger","Italian Salad","Pasta","Risotto","Lasagna","Spaghetti Carbonara","Tiramisu","Toscana","Olio","Lombardy",
    "Emilia-Romagna","Lazio","Cordino","Sanbitter","Chinotto", "Cedrata","Lemonsoda"]

Price = (35, 30, 40, 46, 50, 60, 55, 57, 37,45, 48, 59, 60, 47, 39, 30, 52)

# Customer Details
Customer = {
    "Name": input("Enter Name: "),
    "Contact Number": input("Phone Number: "),
    "Location": input("Location: ")
}

# Display Food Menu
print("\nFOOD MENU")

for i, Food in enumerate(Food_item, start=1):
    print(i, Food, "$", Price[i - 1])

# Get Order
Choice = int(input("\nEnter Food No: "))
Quantity = int(input("Enter the Quantity: "))

# Selected Food
Food = Food_item[Choice - 1]
rate = Price[Choice - 1]

# Calculate Total
Total = rate * Quantity

# Discount
if Total >= 500:
    discount = Total * 0.10
elif Total >= 300:
    discount = Total * 0.05
else:
    discount = 0

# Final Amount
Final_amount = Total - discount

# Order Dictionary
Order = {
    "Customer": Customer["Name"],
    "Contact Number": Customer["Contact Number"],
    "Location": Customer["Location"],
    "Food": Food,
    "Quantity": Quantity,
    "Rate": rate,
    "Total": Total,
    "Discount": discount,
    "Final Amount": Final_amount
}

# Food Bill
print("\nFOOD BILL")

print("Customer Name:", Order["Customer"])
print("Contact Number:", Order["Contact Number"])
print("Location:", Order["Location"])

print("--------------------------------")

print("Food Order:", Order["Food"])
print("Rate:", "$", Order["Rate"])
print("Quantity:", Order["Quantity"])
print("Total:", "$", Order["Total"])
print("Discount:", "$", Order["Discount"])
print("Final Amount:", "$", Order["Final Amount"])
```
